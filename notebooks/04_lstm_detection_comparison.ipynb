# Phase 3: LSTM and classical detector comparison

This notebook reads the latest event-level report produced by `opspilot eval detection`. It does not retrain models or modify the report.

```python
from pathlib import Path

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

report_dir = Path('evals/reports')
if not report_dir.exists():
    report_dir = Path('../evals/reports')
report_paths = sorted(report_dir.glob('detection_results_*.csv'))
if not report_paths:
    raise FileNotFoundError('Run `uv run opspilot eval detection` first.')

results = pd.read_csv(report_paths[-1])
overall = results.loc[results['fault_type'] == 'overall'].copy()
overall[['detector', 'fault_events', 'detected_events', 'false_alarms', 'precision', 'recall', 'f1', 'median_detection_delay_s', 'false_alarms_per_day']]
```

```python
plot_data = overall.melt(
    id_vars='detector',
    value_vars=['f1', 'recall', 'precision'],
    var_name='metric',
    value_name='score',
)

sns.set_theme(style='whitegrid')
axis = sns.barplot(data=plot_data, x='detector', y='score', hue='metric')
axis.set(title='Event-level detection quality', xlabel='', ylabel='Score')
axis.set_ylim(0, 1)
axis.tick_params(axis='x', rotation=15)
plt.tight_layout()
plt.show()
```

```python
sns.barplot(data=overall, x='detector', y='false_alarms_per_day')
plt.title('False alarms per simulated day')
plt.xlabel('')
plt.ylabel('False alarms / day')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
